```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

# Data Loading
Cellphone_test = pd.read_csv("CellPhone_test.csv")
Cellphone_test.head()
```

```python
CellPhone_train = pd.read_csv("CellPhone_train.csv")
CellPhone_train.head()
```

```python
# Latihan data cleaning dan transformasi
CellPhone_train.info() # menampilkan ringkaasan informasi dataset
```

```python
CellPhone_train.describe(include="all") # Menampilkan statistik deskriptif dari dataset
```

```python
# Memeriksa jumlah nilai yang hilang di setiap kolom
missing_values = CellPhone_train.isnull().sum()
missing_values[missing_values > 0]
```

```python
less = missing_values[missing_values < 1].index # menampilkan kolom yang memiliki nilai yang hilang
over = missing_values[missing_values >= 1].index # menampilkan kolom yang memiliki nilai yang hilang


print("--- Kolom dengan missing value < 1 ---")
display(less)

print("\n--- Kolom dengan missing value >= 1 ---")
display(over)
```

```python
#isi nilai numerik yg hilang dengn median
numeric_features = CellPhone_train.select_dtypes(include=['number']).columns
CellPhone_train[numeric_features] = CellPhone_train[numeric_features].fillna(CellPhone_train[numeric_features].median())
CellPhone_train[numeric_features].isnull().sum()
```

```python
# mengisi nilai kategorikal yang hilang dengan modus
categorical_features = CellPhone_train.select_dtypes(include=['object']).columns
for col in categorical_features:
    CellPhone_train[col] = CellPhone_train[col].fillna(CellPhone_train[col].mode()[0])

print("---sisa missing value---")
CellPhone_train[categorical_features].isnull().sum()
```

```python
#hapus kolom yang banyak missing valuenya
df = CellPhone_train.drop(columns=over)
```

```python
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Visualisasi boxplot
for feature in numeric_features:
    plt.figure(figsize=(8,4))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot Of {feature}')
    plt.show()
```

```python
# Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1

# Filter dataframe untuk hanya menyimpan baris yang tidak mengandung outliers pada kolom numerik
condition = ~((df[numeric_features] < (Q1 - 1.5 *IQR)) | (df[numeric_features] > (Q3 + 1.5 * IQR))).any(axis=1)
df_filtered_numeric = df.loc[condition, numeric_features]

# Menggabungkan kembali dengan kolom kategorikal
categorical_features = df.select_dtypes(include=['object']).columns
df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)
```

```python
# Ambil nama kolom yang nilai uniknya lebih dari 3
continuous_features = [col for col in df.columns if df[col].nunique() > 3 and col != 'price_range']

print("Kolom Numerik Kontinu:")
print(continuous_features)
```

```python
#isi outlier dengan median
continuous_features = ['battery_power', 'clock_speed', 'fc', 'int_memory', 'm_dep', 'mobile_wt', 'n_cores', 'pc', 'px_height', 'px_width', 'ram', 'sc_h', 'sc_w', 'talk_time', 'three_g', 'touch_screen', 'wifi']
for col in continuous_features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    median = df[col].median()
df[col] = df[col].apply(lambda x: median if (x < (Q1 - 1.5 * IQR)) or (x > (Q3 + 1.5 * IQR)) else x)
```

```python
# standarisasi fitur numerik
scaler = StandardScaler()
df[continuous_features] = scaler.fit_transform(df[continuous_features])
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(CellPhone_train[continuous_features[3]], kde=True)
plt.title("Histogram Sebelum Standardisasi")

plt.subplot(1, 2, 2)
sns.histplot(df[continuous_features[3]], kde=True)
plt.title("Histogram Setelah Standardisasi")
```

```python
ada_duplikat = df.duplicated().any()
print("Ada duplikat:", ada_duplikat)
```

```python
jumlah = df.duplicated().sum()
print("Jumlah baris duplikat:", jumlah)
```

```python
# Menghapus baris duplikat
df = df.drop_duplicates()

print("DataFrame setelah menghapus duplikat:")
print(df)
```
